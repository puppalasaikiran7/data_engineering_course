```python
class Person:
    def __init__(self,name,age):
        self.name = name 
        self.age = age 

    def introduce(self):
        return f" Hi , iam {self.name} , {self.age} "

class Student(Person):
    def __init__(self,name,age):
        super().__init__(name,age)
        self.marks = []

    def marks_insert(self,*items):
        self.marks.extend(items)
        return self.marks

    def introduce(self):
        average = sum(self.marks)/len(self.marks)
        return f"{super().introduce()} .my average marks are {average}"
        


p = Person("alice",33)
s = Student("saikiran",23)

print(p.name)
print(p.introduce())
print(s.name)
print(s.marks_insert(*[93,95,97]))
print(s.introduce())
```
